# Path 1, P2a — step sizes (Oct 25, 2026)

Data: `load_runs("behavior")`: train groups only (split G), Full, Hierarchy, Star; aborted 1-row rounds and whole-second files dropped. Steps: k = round(10 s). For every row (from the second) and player, |Δk| = size of the move in that row. Bot seats are skipped, as in `run_events` (`src/slider/reversals.py`).

```python
import pandas as pd
from slider import slider_logs as sl
from slider import reversals

runs, table = sl.load_runs("behavior")
pd.crosstab(table["dataset"], table["topology"], margins=True)
```

## 1. Step sizes

Number of player-rows with |Δk| = 1, 2, 3, … (rows with Δk = 0 left out), then the same as % of moves.

```python
counts = reversals.step_counts(runs, table)
counts
```

```python
(100 * counts / counts.sum()).round(2)
```

## 2. More than one player moving in the same row

A player moves in a row if |Δk| ≥ 1. % of all rows and % of rows where at least one player moves.

```python
reversals.simultaneous_moves(runs, table).round(2)
```
