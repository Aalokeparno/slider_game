# Path 1, P2c — reversal and probe counts (Oct 25, 2026)

Data: `load_runs("behavior")`: train groups only (split G), Full, Hierarchy, Star. Events: `detect_events` (m = 2) on every human player. Primary: CC runs first go through `lab_log` (tick 0.1 s), so they are counted like the lab logs. Rate = events / human player-minutes, pooled over the runs of a cell; a round lasts until `fraught.round_end`. 95% CI: group bootstrap (log files, 10,000 resamples, seed 573). Functions: `src/slider/labclock.py`. Nothing is saved.

```python
from slider import slider_logs as sl
from slider import labclock as lc

runs, table = sl.load_runs("behavior")
expo = lc.exposure(runs)
ev = lc.events_table(runs)                      # primary: m = 2, CC through lab_log(tick=0.1)
raw = lc.events_table(runs, lab_clock=False)    # CC on its raw rows
```

## a) Reversals and probes per human player per minute

```python
lc.rates(ev, expo).round(3)
```

## b) CC only: raw vs. lab-logged counts

Share lost = 1 − logged / raw.

```python
lc.logging_loss(raw, ev).round(3)
```

## c) Sensitivity: m = 3 (tick 0.1 s), tick 0.05 s (m = 2)

Point rates per human player per minute. The tick only changes the CC rows.

```python
lc.compare_rates({"primary": ev,
                  "m=3": lc.events_table(runs, m=3),
                  "tick 0.05": lc.events_table(runs, tick=0.05)}, expo).round(3)
```

## d) Share of events that are unresolved (primary)

```python
lc.unresolved_share(ev).round(2)
```
