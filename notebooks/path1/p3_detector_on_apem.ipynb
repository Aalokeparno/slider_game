# Path 1, P3 — the reversal detector on APEM runs (Oct 25, 2026)

Simulation only, no human data. APEM a_θ = s_θ = 0.5, μ = 0.1, speed 1, T = 300 s; 100 runs per topology, seed 573 (one generator per topology).

- **Truth:** `true_events` on each agent's exact path: a move against the direction is a reversal if it travels ≥ 0.2 units, a probe if the agent turns back sooner. Event time = flip time.
- **Detected:** `detect_events` (m = 2 grid steps) on the rows a log keeps: **lab** = `lab_log` (tick 0.1 s, primary), **cloud** = every row where some round(10 s) changes, at exact times.
- **Matching** (`score`): each detected reversal goes to the closest unmatched true reversal within ±0.2 s. Probe fate: detected as a probe, as a reversal that matched no true reversal, or nothing.
- **Sensitivity:** lab with tick 0.05 s; lab with m = 3 (truth then uses 0.3 units).

Functions: `src/slider/apem_truth.py`. Nothing is saved. Humans' logged probe/reversal ratio is about 0.25.
